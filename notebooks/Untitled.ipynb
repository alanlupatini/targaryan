```python
import pandas as pd
import numpy as np

print("Creating Tableau data file...")

# Step 1: Load the 3 cleaned files
df_demo = pd.read_csv('data/Clean/df_demo_clean.csv')
df_web = pd.read_csv('data/Clean/df_web_clean.csv')
df_experiment = pd.read_csv('data/Clean/df_experiment_clean.csv')

print(f"Loaded {len(df_demo)} clients from demo")
print(f"Loaded {len(df_web)} web interactions")
print(f"Loaded {len(df_experiment)} experiment participants")

# Step 2: Merge experiment with demographics
df = df_experiment.merge(df_demo, on='client_id', how='left')
print(f"After merging experiment + demo: {len(df)} rows")

# Step 3: Merge with web behavior
df = df.merge(df_web, on='client_id', how='left')
print(f"After merging with web data: {len(df)} rows")

# Step 4: Create age groups
df['age_group'] = pd.cut(
    df['client_age'],
    bins=[0, 30, 50, 100],
    labels=['Young (<=30)', 'Middle (30-50)', 'Senior (>50)']
)
print("Age groups created")
print(df['age_group'].value_counts())

# Step 5: Create tenure groups
df['tenure_group'] = pd.cut(
    df['client_tenure_years'],
    bins=[0, 6, 11, 16, 100],
    labels=['<6 years', '6-11 years', '11-16 years', '>16 years']
)
print("Tenure groups created")
print(df['tenure_group'].value_counts())

# Step 6: Mark who completed the process
df['completed'] = (df['process_step'] == 'confirm').astype(int)
print(f"Completion marked")
print(f"Total interactions: {len(df)}")
print(f"Confirm interactions: {df['completed'].sum()}")

# Step 7: Basic statistics
print("\nQuick stats:")
print(f"Unique clients: {df['client_id'].nunique()}")
print(f"Control group: {(df['variation'] == 'Control').sum()} interactions")
print(f"Test group: {(df['variation'] == 'Test').sum()} interactions")

# Step 8: Preview
print("\nPreview of final data:")
print(df[['client_id', 'variation', 'age_group', 'gender', 'process_step', 'completed']].head(10))

print("\nAll columns:")
print(df.columns.tolist())

# Step 9: Export for Tableau
output_file = 'data/tableau_complete_data.csv'
df.to_csv(output_file, index=False)

print(f"\nSUCCESS")
print(f"File created: {output_file}")
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(f"\nNext: Import this file into Tableau")
```

```python
print("\nChecking for NULL values:")
print(f"Rows with NULL age_group: {df['age_group'].isna().sum()}")
print(f"Rows with NULL client_age: {df['client_age'].isna().sum()}")
print(f"Rows with NULL gender: {df['gender'].isna().sum()}")

# Check how the merge worked
print("\nMerge quality check:")
print(f"Total rows: {len(df)}")
print(f"Rows with demographic data: {df['client_age'].notna().sum()}")
print(f"Percentage with demographics: {df['client_age'].notna().sum() / len(df) * 100:.2f}%")

# Show age_group distribution
print("\nAge group distribution (excluding NULL):")
print(df['age_group'].value_counts(dropna=False))
```
